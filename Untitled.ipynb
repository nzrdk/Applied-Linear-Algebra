```python
import numpy as np

def compute_eigen_and_verify(matrix):
    if matrix.shape[0] != matrix.shape[1]:
        raise ValueError("Matrix must be square.")
        
    eigenvalues, eigenvectors = np.linalg.eig(matrix)
    
    eigenvalues = np.real(eigenvalues)
    eigenvectors = np.real(eigenvectors)
    
    print("Eigenvalues:\n", eigenvalues)
    print("\nEigenvectors:\n", eigenvectors)
    
    print("\nChecking A * v = lambda * v:")
    for i in range(len(eigenvalues)):
        lam = eigenvalues[i]
        v = eigenvectors[:, i]
        
        left_side = np.dot(matrix, v)
        right_side = lam * v
        
        is_correct = np.allclose(left_side, right_side)
        print(f"For eigenvalue {lam:.4f}: {'Match' if is_correct else 'Mismatch'}")
        
    return eigenvalues, eigenvectors

A = np.array([[4, -2], [1, 1]])
compute_eigen_and_verify(A)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread
from sklearn.decomposition import PCA

image_raw = imread("silcat.jpg")
print("Original image shape:", image_raw.shape)

image_sum = image_raw.sum(axis=2)
print("B&W image shape:", image_sum.shape)
image_bw = image_sum / image_sum.max()
print("Max pixel value:", image_bw.max())

pca = PCA()
pca.fit(image_bw)

cumulative_variance = np.cumsum(pca.explained_variance_ratio_)
components_95 = np.argmax(cumulative_variance >= 0.95) + 1
print(f"Number of components explaining 95% variance: {components_95}")

plt.figure(figsize=(10, 5))
plt.plot(cumulative_variance)
plt.axhline(y=0.95, color='r', linestyle='--', label='95% variance')
plt.axvline(x=components_95, color='k', linestyle='--', label=f'{components_95} components')
plt.title('Cumulative Explained Variance explained by the components')
plt.xlabel('Principal components')
plt.ylabel('Cumulative Explained variance')
plt.legend()
plt.show()

pca_95 = PCA(n_components=components_95)
image_bw_transformed = pca_95.fit_transform(image_bw)
image_bw_reconstructed = pca_95.inverse_transform(image_bw_transformed)

plt.figure(figsize=(8, 6))
plt.imshow(image_bw_reconstructed, cmap='gray')
plt.title(f"Reconstructed image (95% variance, {components_95} components)")
plt.show()

components_list = [5, 15, 25, 75, 100, 170]
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

for ax, n_comp in zip(axes.flatten(), components_list):
    pca_n = PCA(n_components=n_comp)
    img_transformed = pca_n.fit_transform(image_bw)
    img_reconstructed = pca_n.inverse_transform(img_transformed)
    
    ax.imshow(img_reconstructed, cmap='gray')
    ax.set_title(f"Components: {n_comp}")
    ax.axis('on')

plt.tight_layout()
plt.show()
```

```python
import numpy as np

def encrypt_message(message, key_matrix):
    message_vector = np.array([ord(char) for char in message])
    eigenvalues, eigenvectors = np.linalg.eig(key_matrix)
    
    diagonalized_key_matrix = np.dot(
        np.dot(eigenvectors, np.diag(eigenvalues)), 
        np.linalg.inv(eigenvectors)
    )
    
    encrypted_vector = np.dot(diagonalized_key_matrix, message_vector)
    return encrypted_vector

def decrypt_message(encrypted_vector, key_matrix):
    eigenvalues, eigenvectors = np.linalg.eig(key_matrix)
    
    diagonalized_key_matrix = np.dot(
        np.dot(eigenvectors, np.diag(eigenvalues)), 
        np.linalg.inv(eigenvectors)
    )
    
    inv_key_matrix = np.linalg.inv(diagonalized_key_matrix)
    decrypted_vector = np.dot(inv_key_matrix, encrypted_vector)
    decrypted_vector_real = np.real(decrypted_vector)
    
    decrypted_message = "".join([chr(int(round(val))) for val in decrypted_vector_real])
    return decrypted_message

message = "Hello, World!"
key_matrix = np.random.randint(0, 256, (len(message), len(message)))

print("Original Message:", message)

enc_vector = encrypt_message(message, key_matrix)
print("\nEncrypted Message:", enc_vector)

dec_message = decrypt_message(enc_vector, key_matrix)
print("\nDecrypted Message:", dec_message)
```
