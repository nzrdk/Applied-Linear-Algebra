```python
import numpy as np

def compute_and_verify_eigen(A):
    evals, evecs = np.linalg.eig(A)
    evals, evecs = np.real(evals), np.real(evecs)
    
    print("Eigenvalues:\n", evals)
    print("Eigenvectors:\n", evecs)
    
    for i in range(len(evals)):
        v = evecs[:, i]
        is_valid = np.allclose(np.dot(A, v), evals[i] * v)
        print(f"Verify {evals[i]:.4f}: {is_valid}")
        
    return evals, evecs
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread
from sklearn.decomposition import PCA

img_raw = imread("photo.jpg")
print("Raw shape:", img_raw.shape)

img_sum = img_raw.sum(axis=2)
print("BW shape:", img_sum.shape)
img_bw = img_sum / img_sum.max()
print("BW max:", img_bw.max())

pca = PCA().fit(img_bw)
cum_var = np.cumsum(pca.explained_variance_ratio_) * 100
n_95 = np.argmax(cum_var >= 95.0) + 1
print(f"Number of components explaining 95% variance: {n_95}")

plt.plot(cum_var)
plt.axhline(95, c='r', ls='--')
plt.axvline(n_95, c='k', ls='--')
plt.show()

pca_95 = PCA(n_components=n_95)
plt.imshow(pca_95.inverse_transform(pca_95.fit_transform(img_bw)), cmap='gray')
plt.show()

fig, axes = plt.subplots(2, 3, figsize=(12, 8))
for ax, n in zip(axes.ravel(), [5, 15, 25, 75, 100, 170]):
    pca_n = PCA(n_components=n)
    ax.imshow(pca_n.inverse_transform(pca_n.fit_transform(img_bw)), cmap='gray')
    ax.set_title(f"Components: {n}")
plt.show()
```

```python
import numpy as np

def encrypt_message(message, key_matrix):
    msg_vec = np.array([ord(c) for c in message])
    evals, evecs = np.linalg.eig(key_matrix)
    diag_key = np.dot(np.dot(evecs, np.diag(evals)), np.linalg.inv(evecs))
    return np.dot(diag_key, msg_vec)

def decrypt_message_(encrypted_vector, key_matrix):
    evals, evecs = np.linalg.eig(key_matrix)
    inv_diag_key = np.dot(np.dot(evecs, np.diag(1 / evals)), np.linalg.inv(evecs))
    dec_vec = np.dot(inv_diag_key, encrypted_vector)
    return "".join([chr(int(np.round(np.real(v)))) for v in dec_vec])

msg = "Hello, World!"
key = np.random.randint(0, 256, (len(msg), len(msg)))

enc = encrypt_message(msg, key)
print("Encrypted:", enc)
print("Decrypted:", decrypt_message_(enc, key))
```
